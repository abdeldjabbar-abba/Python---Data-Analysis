## Calculate Projected salary next year.
### senior roles assume 5% 
### other roles assume 3%

```python
df_salary_notna = df[pd.notna(df['salary_year_avg'])].copy()
```

```python
def projected_salary_2(row):
    if 'Senior' in row['job_title_short']:
        return 1.05 * row['salary_year_avg']
    else: 
        return 1.03 * row['salary_year_avg']

df_salary_notna['salary_year_inflated'] = df_salary_notna.apply(projected_salary_2, axis=1)

df_salary_notna[['job_title_short', 'salary_year_avg', 'salary_year_inflated']].head(100)
```
