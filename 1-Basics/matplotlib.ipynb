```python
from datasets import load_dataset
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

df['job_posted_date'] = pd.to_datetime(df.job_posted_date)
```

```python
df['job_posted_month'] = df['job_posted_date'].dt.month
```

```python
date_counts =  df.job_posted_month.value_counts().sort_index()
```

```python
plt.plot(date_counts.index, date_counts)
```

```python
jobs_count = df['job_title_short'].value_counts().sort_values(ascending=False)
plt.bar(jobs_count.index, jobs_count)
plt.title('(Bar Chart) of posting by job title')
plt.ylabel('Jobs ')
plt.xlabel('Post counting')
plt.xticks(rotation = 45)
plt.show()
```
